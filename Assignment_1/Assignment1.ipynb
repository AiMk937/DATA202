```python
import numpy as np

data = np.array([42, 45, 47, 50, 210])

mean = np.mean(data)
median = np.median(data)
sample_std = np.std(data, ddof=1)
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

print("Mean:", mean)
print("Median:", median)
print("Sample Standard Deviation:", sample_std)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```
